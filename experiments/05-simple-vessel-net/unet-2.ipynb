```python
import os
os.environ['KMP_DUPLICATE_LIB_OK'] = 'True'
os.environ['CUDA_LAUNCH_BLOCKING'] = '1'

import warnings
warnings.filterwarnings('ignore')

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image
from pathlib import Path
import random
import time
import gc

# Импорты PyTorch
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from pycocotools.coco import COCO

print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")

# =============================================
# ПРОСТАЯ И ЭФФЕКТИВНАЯ КОНФИГУРАЦИЯ
# =============================================

IMG_SIZE = (256, 256)  # Вернули к нормальному размеру
BATCH_SIZE = 16
EPOCHS = 50
LEARNING_RATE = 1e-4
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')

BASE_DIR = Path("archive/coco2017")
TRAIN_IMAGE_DIR = BASE_DIR / "filtered_train2017"
VAL_IMAGE_DIR = BASE_DIR / "filtered_val2017"
TRAIN_ANNOTATION = BASE_DIR / "annotations/filtered_train_annotations.json"
VAL_ANNOTATION = BASE_DIR / "annotations/filtered_val_annotations.json"

# =============================================
# ПРОСТАЯ И СТАБИЛЬНАЯ МОДЕЛЬ
# =============================================

class SimpleVesselNet(nn.Module):
    def __init__(self):
        super().__init__()
        
        # Encoder
        self.enc1 = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),
            nn.Conv2d(32, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True)
        )
        
        self.enc2 = nn.Sequential(
            nn.Conv2d(32, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.Conv2d(64, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True)
        )
        
        self.enc3 = nn.Sequential(
            nn.Conv2d(64, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.Conv2d(128, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True)
        )
        
        # Bottleneck
        self.bottleneck = nn.Sequential(
            nn.Conv2d(128, 256, 3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.Conv2d(256, 256, 3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.Conv2d(256, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True)
        )
        
        # Decoder
        self.dec1 = nn.Sequential(
            nn.Conv2d(256, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.Conv2d(128, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True)
        )
        
        self.dec2 = nn.Sequential(
            nn.Conv2d(128, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.Conv2d(64, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True)
        )
        
        self.dec3 = nn.Sequential(
            nn.Conv2d(64, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),
            nn.Conv2d(32, 16, 3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(inplace=True)
        )
        
        # Output
        self.output = nn.Sequential(
            nn.Conv2d(16, 8, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(8, 1, 1)
        )
        
        self.pool = nn.MaxPool2d(2)
        self.upsample = nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True)
        
    def forward(self, x):
        # Encoder
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool(e1))
        e3 = self.enc3(self.pool(e2))
        
        # Bottleneck
        b = self.bottleneck(self.pool(e3))
        
        # Decoder с skip connections
        d1 = self.upsample(b)
        d1 = torch.cat([d1, e3], dim=1)
        d1 = self.dec1(d1)
        
        d2 = self.upsample(d1)
        d2 = torch.cat([d2, e2], dim=1)
        d2 = self.dec2(d2)
        
        d3 = self.upsample(d2)
        d3 = torch.cat([d3, e1], dim=1)
        d3 = self.dec3(d3)
        
        return self.output(d3)

# =============================================
# СТАБИЛЬНАЯ ФУНКЦИЯ ПОТЕРЬ
# =============================================

class StableVesselLoss(nn.Module):
    def __init__(self, pos_weight=8.0):
        super().__init__()
        self.pos_weight = pos_weight
        
    def forward(self, pred, target):
        # Простая BCE с умеренным весом
        bce_loss = F.binary_cross_entropy_with_logits(
            pred, target, 
            pos_weight=torch.tensor([self.pos_weight]).to(pred.device)
        )
        
        # Простой Dice loss
        pred_sigmoid = torch.sigmoid(pred)
        intersection = (pred_sigmoid * target).sum()
        union = pred_sigmoid.sum() + target.sum()
        dice_loss = 1 - (2. * intersection + 1e-6) / (union + 1e-6)
        
        # Простая комбинация
        total_loss = bce_loss + dice_loss
        
        return total_loss

# =============================================
# ДАТАСЕТ С ПРЕДЗАГРУЗКОЙ ДЛЯ СКОРОСТИ
# =============================================

class CachedVesselDataset(Dataset):
    def __init__(self, coco, img_dir, transform=None, max_samples=1000):
        self.coco = coco
        self.img_dir = img_dir
        self.transform = transform
        
        self.ids = list(coco.imgs.keys())[:max_samples]
        print(f"📊 Загружаем {len(self.ids)} изображений")
        
        # ПРЕДЗАГРУЗКА ВСЕХ ДАННЫХ
        self.data = []
        print("🔄 Предзагрузка данных...")
        
        for i, img_id in enumerate(self.ids):
            if i % 100 == 0:
                print(f"   Загружено {i}/{len(self.ids)}")
                
            img_info = coco.loadImgs(img_id)[0]
            img_path = self.img_dir / img_info['file_name']
            
            img = Image.open(img_path).convert('RGB')
            
            # Маска
            ann_ids = coco.getAnnIds(imgIds=img_id)
            anns = coco.loadAnns(ann_ids)
            mask = np.zeros((img_info['height'], img_info['width']), dtype=np.uint8)
            
            for ann in anns:
                ann_mask = coco.annToMask(ann)
                mask = np.maximum(mask, ann_mask)
            
            mask = Image.fromarray(mask)
            
            if self.transform:
                img_tensor = self.transform(img)
                mask_tensor = transforms.Resize(IMG_SIZE, Image.NEAREST)(mask)
                mask_tensor = transforms.ToTensor()(mask_tensor)
                mask_tensor = (mask_tensor > 0).float()
            
            self.data.append((img_tensor, mask_tensor))
        
        print("✅ Все данные загружены в память!")
    
    def __len__(self):
        return len(self.data)
    
    def __getitem__(self, idx):
        return self.data[idx]

# =============================================
# ПРОСТЫЕ ТРАНСФОРМЫ
# =============================================

train_transform = transforms.Compose([
    transforms.Resize(IMG_SIZE),
    transforms.RandomHorizontalFlip(0.5),
    transforms.ToTensor(),
])

val_transform = transforms.Compose([
    transforms.Resize(IMG_SIZE),
    transforms.ToTensor(),
])

# =============================================
# ОБУЧЕНИЕ С ГРАДУАЛЬНЫМ УСЛОЖНЕНИЕМ
# =============================================

def main():
    print("🚀 ЗАПУСК СТАБИЛЬНОГО ОБУЧЕНИЯ...")
    
    # Загружаем аннотации
    print("📁 Загрузка аннотаций...")
    train_coco = COCO(TRAIN_ANNOTATION)
    val_coco = COCO(VAL_ANNOTATION)
    
    # Создаем датасеты с предзагрузкой
    print("🔄 Создание датасетов...")
    train_dataset = CachedVesselDataset(train_coco, TRAIN_IMAGE_DIR, 
                                      transform=train_transform, max_samples=1000)
    val_dataset = CachedVesselDataset(val_coco, VAL_IMAGE_DIR, 
                                    transform=val_transform, max_samples=200)
    
    train_loader = DataLoader(
        train_dataset,
        batch_size=BATCH_SIZE,
        shuffle=True,
        num_workers=0,
        pin_memory=False
    )
    
    val_loader = DataLoader(
        val_dataset,
        batch_size=BATCH_SIZE,
        shuffle=False,
        num_workers=0,
        pin_memory=False
    )
    
    print(f"📊 Данные готовы: {len(train_dataset)} train, {len(val_dataset)} val")
    
    # Модель
    model = SimpleVesselNet().to(DEVICE)
    
    # Проверка модели
    print("🧪 Тест модели...")
    test_input = torch.randn(2, 3, 256, 256).to(DEVICE)
    with torch.no_grad():
        test_output = model(test_input)
        test_sigmoid = torch.sigmoid(test_output)
        print(f"✅ Модель работает! Output range: [{test_sigmoid.min().item():.3f}, {test_sigmoid.max().item():.3f}]")
    
    # Функция потерь и оптимизатор
    criterion = StableVesselLoss(pos_weight=8.0)
    optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE, weight_decay=1e-5)
    
    # Простой планировщик
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=15, gamma=0.5)
    
    print(f"🧠 Модель создана: {sum(p.numel() for p in model.parameters()):,} параметров")
    
    # ОБУЧЕНИЕ
    print(f"\n🎯 НАЧИНАЕМ СТАБИЛЬНОЕ ОБУЧЕНИЕ...")
    print("=" * 60)
    
    best_dice = 0.0
    patience = 0
    max_patience = 20
    
    for epoch in range(EPOCHS):
        epoch_start = time.time()
        
        # ОБУЧЕНИЕ
        model.train()
        train_loss = 0
        train_batches = 0
        
        print(f"\n📍 Эпоха {epoch+1}/{EPOCHS}")
        
        for batch_idx, (images, masks) in enumerate(train_loader):
            images = images.to(DEVICE)
            masks = masks.to(DEVICE)
            
            # Forward pass
            outputs = model(images)
            loss = criterion(outputs, masks)
            
            # Backward pass
            optimizer.zero_grad()
            loss.backward()
            
            # Gentle gradient clipping
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=0.5)
            optimizer.step()
            
            train_loss += loss.item()
            train_batches += 1
            
            # Мониторинг с акцентом на то, что модель ВООБЩЕ предсказывает
            if batch_idx % 20 == 0:
                with torch.no_grad():
                    pred_sigmoid = torch.sigmoid(outputs)
                    pred_pixels = (pred_sigmoid > 0.1).float().mean().item()  # Низкий порог для начала
                    true_pixels = masks.mean().item()
                    
                    # Проверяем что модель не "мертвая"
                    output_range = f"[{pred_sigmoid.min().item():.3f}, {pred_sigmoid.max().item():.3f}]"
                    
                    print(f"   Батч {batch_idx:3d} | Loss: {loss.item():.4f} | "
                          f"Pred: {pred_pixels:.4f} | True: {true_pixels:.4f} | "
                          f"Range: {output_range}")
        
        avg_train_loss = train_loss / train_batches
        scheduler.step()
        current_lr = optimizer.param_groups[0]['lr']
        
        # ВАЛИДАЦИЯ
        model.eval()
        val_loss = 0
        val_dice = 0
        val_recall = 0
        val_batches = 0
        
        with torch.no_grad():
            for images, masks in val_loader:
                images = images.to(DEVICE)
                masks = masks.to(DEVICE)
                outputs = model(images)
                loss = criterion(outputs, masks)
                
                pred_sigmoid = torch.sigmoid(outputs)
                pred_binary = (pred_sigmoid > 0.3).float()
                
                # Метрики
                tp = (pred_binary * masks).sum()
                fp = (pred_binary * (1 - masks)).sum()
                fn = ((1 - pred_binary) * masks).sum()
                
                precision = tp / (tp + fp + 1e-6)
                recall = tp / (tp + fn + 1e-6)
                dice = (2 * tp) / (2 * tp + fp + fn + 1e-6)
                
                val_loss += loss.item()
                val_dice += dice.item()
                val_recall += recall.item()
                val_batches += 1
        
        avg_val_loss = val_loss / val_batches
        avg_val_dice = val_dice / val_batches
        avg_val_recall = val_recall / val_batches
        
        epoch_time = time.time() - epoch_start
        
        print(f"✅ Эпоха {epoch+1} завершена за {epoch_time:.1f}с")
        print(f"📊 Train Loss: {avg_train_loss:.4f} | Val Loss: {avg_val_loss:.4f}")
        print(f"🎯 Dice: {avg_val_dice:.4f} | Recall: {avg_val_recall:.4f}")
        print(f"📈 LR: {current_lr:.2e}")
        
        # КРИТИЧЕСКАЯ ПРОВЕРКА: модель не должна быть "мертвой"
        if avg_val_dice == 0.0 and epoch > 5:
            print("⚠️  КРИТИЧЕСКАЯ ОШИБКА: Модель не учится! Проверяем...")
            debug_model_outputs(model, val_loader, DEVICE)
            break
        
        # Сохранение лучшей модели
        if avg_val_dice > best_dice:
            best_dice = avg_val_dice
            patience = 0
            torch.save({
                'epoch': epoch,
                'model_state_dict': model.state_dict(),
                'optimizer_state_dict': optimizer.state_dict(),
                'dice': best_dice,
                'recall': avg_val_recall,
            }, 'stable_vessel_model.pth')
            print(f"💾 Новая лучшая модель! Dice: {best_dice:.4f}")
        else:
            patience += 1
            print(f"⏳ Patience: {patience}/{max_patience}")
        
        # Визуализация прогресса
        if (epoch + 1) % 10 == 0 or epoch == 0:
            visualize_stable_progress(model, val_dataset, DEVICE, epoch+1)
        
        # Ранняя остановка
        if patience >= max_patience:
            print(f"🛑 Ранняя остановка на эпохе {epoch+1}")
            break
    
    print(f"\n🎉 ОБУЧЕНИЕ ЗАВЕРШЕНО!")
    print(f"🏆 Лучший Dice: {best_dice:.4f}")

def debug_model_outputs(model, val_loader, device):
    """Диагностика выходов модели"""
    print("🔍 ДИАГНОСТИКА МОДЕЛИ...")
    model.eval()
    
    with torch.no_grad():
        for images, masks in val_loader:
            images = images.to(device)
            outputs = model(images)
            pred_sigmoid = torch.sigmoid(outputs)
            
            print(f"Output range: [{pred_sigmoid.min().item():.3f}, {pred_sigmoid.max().item():.3f}]")
            print(f"Output mean: {pred_sigmoid.mean().item():.3f}")
            print(f"Output std: {pred_sigmoid.std().item():.3f}")
            
            # Проверяем несколько предсказаний
            for i in range(min(3, images.shape[0])):
                single_pred = pred_sigmoid[i]
                print(f"Sample {i}: min={single_pred.min().item():.3f}, max={single_pred.max().item():.3f}, mean={single_pred.mean().item():.3f}")
            
            break

def visualize_stable_progress(model, dataset, device, epoch, num_samples=3):
    """Визуализация прогресса"""
    model.eval()
    
    indices = random.sample(range(len(dataset)), num_samples)
    
    fig, axes = plt.subplots(num_samples, 3, figsize=(12, 4 * num_samples))
    
    for i, idx in enumerate(indices):
        with torch.no_grad():
            image, true_mask = dataset[idx]
            pred = model(image.unsqueeze(0).to(device))
            pred_sigmoid = torch.sigmoid(pred)
            pred_binary = (pred_sigmoid > 0.3).float()
            
            true_pixels = true_mask.sum().item()
            pred_pixels = pred_binary.sum().item()
            
            # Dice score
            intersection = (pred_binary * true_mask.to(device)).sum()
            union = pred_binary.sum() + true_mask.to(device).sum()
            dice = (2. * intersection / (union + 1e-6)).item()
        
        # Изображение
        img_np = image.cpu().permute(1, 2, 0).numpy()
        img_np = np.clip(img_np, 0, 1)
        
        axes[i, 0].imshow(img_np)
        axes[i, 0].set_title('Image')
        axes[i, 0].axis('off')
        
        # True mask
        true_np = true_mask.squeeze().cpu().numpy()
        axes[i, 1].imshow(true_np, cmap='gray')
        axes[i, 1].set_title(f'True: {true_pixels}px')
        axes[i, 1].axis('off')
        
        # Prediction
        pred_np = pred_binary.squeeze().cpu().numpy()
        axes[i, 2].imshow(pred_np, cmap='gray')
        axes[i, 2].set_title(f'Pred: {pred_pixels}px\nDice: {dice:.3f}')
        axes[i, 2].axis('off')
    
    plt.suptitle(f'Epoch {epoch} - Stable Training', fontsize=16)
    plt.tight_layout()
    plt.savefig(f'stable_progress_epoch_{epoch}.png', dpi=100, bbox_inches='tight')
    plt.show()

# =============================================
# ЗАПУСК
# =============================================

if __name__ == "__main__":
    print("⚡ STABLE VESSEL TRAINING - ПРОСТАЯ И ЭФФЕКТИВНАЯ АРХИТЕКТУРА")
    print("=" * 60)
    
    main()
```
