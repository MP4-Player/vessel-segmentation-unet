```python
import os
os.environ['KMP_DUPLICATE_LIB_OK'] = 'True'
os.environ['CUDA_LAUNCH_BLOCKING'] = '1'

import warnings
warnings.filterwarnings('ignore')

import numpy as np
import matplotlib.pyplot as plt
from PIL import Image, ImageDraw
from pathlib import Path
import random
import time
import gc

# Импорты PyTorch
import torch
import torch.nn as nn
import torch.nn.functional as F
import torch.optim as optim
from torch.utils.data import Dataset, DataLoader
from torchvision import transforms
from pycocotools.coco import COCO

print(f"PyTorch version: {torch.__version__}")
print(f"CUDA available: {torch.cuda.is_available()}")

# =============================================
# КОНФИГУРАЦИЯ
# =============================================

IMG_SIZE = (256, 256)
BATCH_SIZE = 16
EPOCHS = 50
LEARNING_RATE = 1e-4
DEVICE = torch.device('cuda' if torch.cuda.is_available() else 'cpu')
MAX_OBJECTS = 10  # Увеличили максимальное количество объектов

BASE_DIR = Path("archive/coco2017")
TRAIN_IMAGE_DIR = BASE_DIR / "filtered_train2017"
VAL_IMAGE_DIR = BASE_DIR / "filtered_val2017"
TRAIN_ANNOTATION = BASE_DIR / "annotations/filtered_train_annotations.json"
VAL_ANNOTATION = BASE_DIR / "annotations/filtered_val_annotations.json"

# =============================================
# ИСПРАВЛЕННЫЙ DATASET С МНОГИМИ BBOXES
# =============================================

class SmartZoomVesselDataset(Dataset):
    def __init__(self, coco, img_dir, transform=None, max_samples=1000, target_size=256, max_objects=MAX_OBJECTS):
        self.coco = coco
        self.img_dir = img_dir
        self.transform = transform
        self.target_size = target_size
        self.max_objects = max_objects
        self.ids = list(coco.imgs.keys())[:max_samples]
        
        print(f"📊 Загружаем {len(self.ids)} изображений с умным зумом")
        
        # Кэшируем информацию об аннотациях для быстрого доступа
        self.annotations_cache = {}
        for img_id in self.ids:
            ann_ids = coco.getAnnIds(imgIds=img_id)
            anns = coco.loadAnns(ann_ids)
            self.annotations_cache[img_id] = anns
    
    def __len__(self):
        return len(self.ids)
    
    def __getitem__(self, idx):
        img_id = self.ids[idx]
        img_info = self.coco.loadImgs(img_id)[0]
        img_path = self.img_dir / img_info['file_name']
        
        # Загружаем изображение
        img = Image.open(img_path).convert('RGB')
        original_width, original_height = img.size
        
        # Создаем маску и находим ВСЕ bounding boxes
        anns = self.annotations_cache[img_id]
        mask = np.zeros((original_height, original_width), dtype=np.uint8)
        boxes = []
        
        for ann in anns:
            ann_mask = self.coco.annToMask(ann)
            mask = np.maximum(mask, ann_mask)
            
            # Получаем bbox из аннотации
            x, y, w, h = ann['bbox']
            boxes.append([x, y, x + w, y + h])  # [x1, y1, x2, y2]
        
        mask = Image.fromarray(mask)
        
        # ВЫЧИСЛЯЕМ УМНЫЙ ZOOM
        img, mask, boxes = self.smart_zoom(img, mask, boxes, original_width, original_height)
        
        # Применяем трансформации
        if self.transform:
            img_tensor = self.transform(img)
        else:
            img_tensor = transforms.ToTensor()(img)
        
        # Масштабируем маску
        mask_resized = mask.resize((self.target_size, self.target_size), Image.NEAREST)
        mask_tensor = transforms.ToTensor()(mask_resized)
        mask_tensor = (mask_tensor > 0).float()
        
        # Масштабируем ВСЕ bboxes относительно нового размера изображения
        scale_x = self.target_size / img.width
        scale_y = self.target_size / img.height
        
        scaled_boxes = []
        for box in boxes:
            x1, y1, x2, y2 = box
            scaled_boxes.append([
                x1 * scale_x, y1 * scale_y, 
                x2 * scale_x, y2 * scale_y
            ])
        
        # Создаем tensor для ВСЕХ bboxes
        bbox_tensor = self.boxes_to_tensor(scaled_boxes, self.target_size)
        
        return img_tensor, mask_tensor, bbox_tensor
    
    def smart_zoom(self, img, mask, boxes, original_width, original_height):
        """Умный зум: чем меньше площадь сосудов, тем больше зум"""
        
        # Вычисляем общую площадь сосудов
        mask_array = np.array(mask)
        vessel_pixels = np.sum(mask_array > 0)
        total_pixels = original_width * original_height
        vessel_coverage = vessel_pixels / total_pixels
        
        # Определяем коэффициент зума на основе покрытия сосудами
        if vessel_coverage < 0.01:  # Очень маленькие сосуды (<1%)
            zoom_factor = random.uniform(2.0, 3.0)
        elif vessel_coverage < 0.03:  # Маленькие сосуды (1-3%)
            zoom_factor = random.uniform(1.5, 2.5)
        elif vessel_coverage < 0.05:  # Средние сосуды (3-5%)
            zoom_factor = random.uniform(1.2, 1.8)
        else:  # Большие сосуды (>5%)
            zoom_factor = random.uniform(1.0, 1.3)
        
        # Находим общий bounding box ВСЕХ сосудов
        if boxes:
            all_x1 = min(box[0] for box in boxes)
            all_y1 = min(box[1] for box in boxes)
            all_x2 = max(box[2] for box in boxes)
            all_y2 = max(box[3] for box in boxes)
        else:
            # Если нет сосудов (маловероятно), используем центр
            all_x1, all_y1, all_x2, all_y2 = (
                original_width * 0.25, original_height * 0.25,
                original_width * 0.75, original_height * 0.75
            )
        
        # Центр зума - центр всех сосудов
        center_x = (all_x1 + all_x2) / 2
        center_y = (all_y1 + all_y2) / 2
        
        # Вычисляем размер области после зума
        new_width = original_width / zoom_factor
        new_height = original_height / zoom_factor
        
        # Область кропа
        x1 = max(0, center_x - new_width / 2)
        y1 = max(0, center_y - new_height / 2)
        x1 = min(x1, original_width - new_width)
        y1 = min(y1, original_height - new_height)
        x2 = x1 + new_width
        y2 = y1 + new_height
        
        # Кропаем изображение и маску
        img_cropped = img.crop((x1, y1, x2, y2))
        mask_cropped = mask.crop((x1, y1, x2, y2))
        
        # Обновляем ВСЕ bounding boxes
        new_boxes = []
        for box in boxes:
            bx1, by1, bx2, by2 = box
            # Проверяем пересечение с кропом
            if (bx2 > x1 and bx1 < x2 and by2 > y1 and by1 < y2):
                new_box = [
                    max(bx1 - x1, 0), max(by1 - y1, 0),
                    min(bx2 - x1, new_width), min(by2 - y1, new_height)
                ]
                new_boxes.append(new_box)
        
        # Ресайзим к обратному размеру
        img_resized = img_cropped.resize((original_width, original_height), Image.BILINEAR)
        mask_resized = mask_cropped.resize((original_width, original_height), Image.NEAREST)
        
        # Масштабируем ВСЕ bounding boxes
        scale_x = original_width / new_width
        scale_y = original_height / new_height
        scaled_boxes = []
        for box in new_boxes:
            x1, y1, x2, y2 = box
            scaled_boxes.append([
                x1 * scale_x, y1 * scale_y,
                x2 * scale_x, y2 * scale_y
            ])
        
        return img_resized, mask_resized, scaled_boxes
    
    def boxes_to_tensor(self, boxes, img_size):
        """Конвертируем ВСЕ bboxes в tensor формата [x_center, y_center, width, height, confidence]"""
        # Создаем тензор для максимального количества объектов
        # Формат: [x_center, y_center, width, height, confidence] × MAX_OBJECTS
        bbox_tensor = torch.zeros(self.max_objects * 5)
        
        for i, box in enumerate(boxes[:self.max_objects]):  # Берем не более MAX_OBJECTS
            x1, y1, x2, y2 = box
            
            x_center = (x1 + x2) / 2 / img_size
            y_center = (y1 + y2) / 2 / img_size
            width = (x2 - x1) / img_size
            height = (y2 - y1) / img_size
            confidence = 1.0  # Для ground truth уверенность всегда 1
            
            # Записываем в тензор
            start_idx = i * 5
            bbox_tensor[start_idx] = x_center
            bbox_tensor[start_idx + 1] = y_center
            bbox_tensor[start_idx + 2] = width
            bbox_tensor[start_idx + 3] = height
            bbox_tensor[start_idx + 4] = confidence
        
        return bbox_tensor

# =============================================
# УПРОЩЕННАЯ MULTI-TASK МОДЕЛЬ
# =============================================

class MultiTaskVesselNet(nn.Module):
    def __init__(self, max_objects=MAX_OBJECTS):
        super().__init__()
        self.max_objects = max_objects
        
        # Общий энкодер
        self.enc1 = nn.Sequential(
            nn.Conv2d(3, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),
            nn.Conv2d(32, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True)
        )
        
        self.enc2 = nn.Sequential(
            nn.Conv2d(32, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.Conv2d(64, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True)
        )
        
        self.enc3 = nn.Sequential(
            nn.Conv2d(64, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.Conv2d(128, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True)
        )
        
        # Bottleneck
        self.bottleneck = nn.Sequential(
            nn.Conv2d(128, 256, 3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True),
            nn.Conv2d(256, 256, 3, padding=1),
            nn.BatchNorm2d(256),
            nn.ReLU(inplace=True)
        )
        
        # Декодер для сегментации
        self.dec1 = nn.Sequential(
            nn.Conv2d(256 + 128, 128, 3, padding=1),
            nn.BatchNorm2d(128),
            nn.ReLU(inplace=True),
            nn.Conv2d(128, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True)
        )
        
        self.dec2 = nn.Sequential(
            nn.Conv2d(64 + 64, 64, 3, padding=1),
            nn.BatchNorm2d(64),
            nn.ReLU(inplace=True),
            nn.Conv2d(64, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True)
        )
        
        self.dec3 = nn.Sequential(
            nn.Conv2d(32 + 32, 32, 3, padding=1),
            nn.BatchNorm2d(32),
            nn.ReLU(inplace=True),
            nn.Conv2d(32, 16, 3, padding=1),
            nn.BatchNorm2d(16),
            nn.ReLU(inplace=True)
        )
        
        # Head для сегментации
        self.segmentation_head = nn.Sequential(
            nn.Conv2d(16, 8, 3, padding=1),
            nn.ReLU(inplace=True),
            nn.Conv2d(8, 1, 1)
        )
        
        # Head для detection (bbox regression для ВСЕХ объектов)
        self.detection_head = nn.Sequential(
            nn.AdaptiveAvgPool2d(1),
            nn.Flatten(),
            nn.Linear(256, 512),
            nn.ReLU(inplace=True),
            nn.Dropout(0.2),
            nn.Linear(512, 256),
            nn.ReLU(inplace=True),
            nn.Linear(256, max_objects * 5)  # [x_center, y_center, width, height, confidence] × max_objects
        )
        
        self.pool = nn.MaxPool2d(2)
        self.upsample = nn.Upsample(scale_factor=2, mode='bilinear', align_corners=True)
        
    def forward(self, x):
        # Encoder
        e1 = self.enc1(x)
        e2 = self.enc2(self.pool(e1))
        e3 = self.enc3(self.pool(e2))
        
        # Bottleneck
        b = self.bottleneck(self.pool(e3))
        
        # Detection head (из bottleneck) - ВСЕ объекты
        bbox_output = self.detection_head(b)
        bbox_output = bbox_output.view(-1, self.max_objects, 5)  # [batch, max_objects, 5]
        
        # Segmentation decoder
        d1 = self.upsample(b)
        d1 = torch.cat([d1, e3], dim=1)
        d1 = self.dec1(d1)
        
        d2 = self.upsample(d1)
        d2 = torch.cat([d2, e2], dim=1)
        d2 = self.dec2(d2)
        
        d3 = self.upsample(d2)
        d3 = torch.cat([d3, e1], dim=1)
        d3 = self.dec3(d3)
        
        segmentation_output = self.segmentation_head(d3)
        
        return segmentation_output, bbox_output

# =============================================
# ИСПРАВЛЕННЫЙ MULTI-TASK LOSS
# =============================================

class MultiTaskLoss(nn.Module):
    def __init__(self, seg_weight=1.0, det_weight=0.5, pos_weight=8.0, max_objects=MAX_OBJECTS):
        super().__init__()
        self.seg_weight = seg_weight
        self.det_weight = det_weight
        self.pos_weight = pos_weight
        self.max_objects = max_objects
        
    def forward(self, seg_pred, bbox_pred, seg_target, bbox_target):
        # Segmentation loss
        seg_loss = F.binary_cross_entropy_with_logits(
            seg_pred, seg_target, 
            pos_weight=torch.tensor([self.pos_weight]).to(seg_pred.device)
        )
        
        pred_sigmoid = torch.sigmoid(seg_pred)
        intersection = (pred_sigmoid * seg_target).sum()
        union = pred_sigmoid.sum() + seg_target.sum()
        dice_loss = 1 - (2. * intersection + 1e-6) / (union + 1e-6)
        
        total_seg_loss = seg_loss + dice_loss
        
        # Detection loss для ВСЕХ объектов
        batch_size = bbox_target.size(0)
        det_loss = 0
        obj_count = 0
        
        for i in range(batch_size):
            for j in range(self.max_objects):
                # Проверяем, есть ли объект на этой позиции
                if bbox_target[i, j, 4] > 0.5:  # confidence > 0.5
                    # Smooth L1 для координат bbox
                    coord_loss = F.smooth_l1_loss(bbox_pred[i, j, :4], bbox_target[i, j, :4])
                    # BCE для confidence (ИСПРАВЛЕНО: правильная размерность)
                    conf_loss = F.binary_cross_entropy_with_logits(
                        bbox_pred[i, j, 4:5],  # [1]
                        bbox_target[i, j, 4:5]  # [1]
                    )
                    det_loss += coord_loss + conf_loss
                    obj_count += 1
        
        # Усредняем по количеству объектов
        if obj_count > 0:
            det_loss = det_loss / obj_count
        else:
            det_loss = torch.tensor(0.0).to(seg_pred.device)
        
        # Combined loss
        total_loss = self.seg_weight * total_seg_loss + self.det_weight * det_loss
        
        return total_loss, total_seg_loss, det_loss

# =============================================
# ОБНОВЛЕННОЕ ОБУЧЕНИЕ
# =============================================

def main():
    print("🚀 ЗАПУСК MULTI-TASK ОБУЧЕНИЯ С ВСЕМИ BBOXES...")
    
    # Загружаем аннотации
    train_coco = COCO(TRAIN_ANNOTATION)
    val_coco = COCO(VAL_ANNOTATION)
    
    # Создаем датасеты с умным зумом
    train_transform = transforms.Compose([
        transforms.Resize(IMG_SIZE),
        transforms.RandomHorizontalFlip(0.5),
        transforms.ColorJitter(0.1, 0.1, 0.1, 0.1),
        transforms.ToTensor(),
    ])
    
    val_transform = transforms.Compose([
        transforms.Resize(IMG_SIZE),
        transforms.ToTensor(),
    ])
    
    train_dataset = SmartZoomVesselDataset(train_coco, TRAIN_IMAGE_DIR, 
                                         transform=train_transform, max_samples=1000)
    val_dataset = SmartZoomVesselDataset(val_coco, VAL_IMAGE_DIR, 
                                       transform=val_transform, max_samples=200)
    
    train_loader = DataLoader(train_dataset, batch_size=BATCH_SIZE, shuffle=True, num_workers=0)
    val_loader = DataLoader(val_dataset, batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
    
    print(f"📊 Данные готовы: {len(train_dataset)} train, {len(val_dataset)} val")
    print(f"🎯 Максимальное количество объектов на изображении: {MAX_OBJECTS}")
    
    # Модель
    model = MultiTaskVesselNet(max_objects=MAX_OBJECTS).to(DEVICE)
    
    # Проверка модели
    print("🧪 Тест модели...")
    test_input = torch.randn(2, 3, 256, 256).to(DEVICE)
    with torch.no_grad():
        seg_output, bbox_output = model(test_input)
        print(f"✅ Модель работает! Seg: {seg_output.shape}, Bbox: {bbox_output.shape}")
        print(f"   Bbox output содержит {MAX_OBJECTS} объектов по 5 параметров каждый")
    
    # Функция потерь и оптимизатор
    criterion = MultiTaskLoss(seg_weight=1.0, det_weight=0.5, pos_weight=8.0, max_objects=MAX_OBJECTS)
    optimizer = optim.Adam(model.parameters(), lr=LEARNING_RATE, weight_decay=1e-5)
    scheduler = optim.lr_scheduler.StepLR(optimizer, step_size=15, gamma=0.5)
    
    print(f"🧠 Модель создана: {sum(p.numel() for p in model.parameters()):,} параметров")
    
    # ОБУЧЕНИЕ
    best_dice = 0.0
    patience = 0
    max_patience = 20
    
    for epoch in range(EPOCHS):
        epoch_start = time.time()
        
        # ОБУЧЕНИЕ
        model.train()
        train_total_loss = 0
        train_seg_loss = 0
        train_det_loss = 0
        train_batches = 0
        
        print(f"\n📍 Эпоха {epoch+1}/{EPOCHS}")
        
        for batch_idx, (images, masks, bboxes) in enumerate(train_loader):
            images = images.to(DEVICE)
            masks = masks.to(DEVICE)
            bboxes = bboxes.view(-1, MAX_OBJECTS, 5).to(DEVICE)  # Reshape для multi-object
            
            # Forward pass
            seg_output, bbox_output = model(images)
            total_loss, seg_loss, det_loss = criterion(seg_output, bbox_output, masks, bboxes)
            
            # Backward pass
            optimizer.zero_grad()
            total_loss.backward()
            torch.nn.utils.clip_grad_norm_(model.parameters(), max_norm=0.5)
            optimizer.step()
            
            train_total_loss += total_loss.item()
            train_seg_loss += seg_loss.item()
            train_det_loss += det_loss.item()
            train_batches += 1
            
            if batch_idx % 20 == 0:
                with torch.no_grad():
                    pred_sigmoid = torch.sigmoid(seg_output)
                    pred_pixels = (pred_sigmoid > 0.1).float().mean().item()
                    true_pixels = masks.mean().item()
                    
                    # Подсчитываем количество обнаруженных объектов
                    obj_count = (bboxes[:, :, 4] > 0.5).sum().item()
                    
                    print(f"   Батч {batch_idx:3d} | Total: {total_loss.item():.4f} | "
                          f"Seg: {seg_loss.item():.4f} | Det: {det_loss.item():.4f} | "
                          f"Pred: {pred_pixels:.4f} | True: {true_pixels:.4f} | "
                          f"Objects: {obj_count}")
        
        # ВАЛИДАЦИЯ
        model.eval()
        val_total_loss = 0
        val_seg_loss = 0
        val_det_loss = 0
        val_dice = 0
        val_batches = 0
        
        with torch.no_grad():
            for images, masks, bboxes in val_loader:
                images = images.to(DEVICE)
                masks = masks.to(DEVICE)
                bboxes = bboxes.view(-1, MAX_OBJECTS, 5).to(DEVICE)
                
                seg_output, bbox_output = model(images)
                total_loss, seg_loss, det_loss = criterion(seg_output, bbox_output, masks, bboxes)
                
                # Метрики сегментации
                pred_sigmoid = torch.sigmoid(seg_output)
                pred_binary = (pred_sigmoid > 0.3).float()
                
                tp = (pred_binary * masks).sum()
                fp = (pred_binary * (1 - masks)).sum()
                fn = ((1 - pred_binary) * masks).sum()
                dice = (2 * tp) / (2 * tp + fp + fn + 1e-6)
                
                val_total_loss += total_loss.item()
                val_seg_loss += seg_loss.item()
                val_det_loss += det_loss.item()
                val_dice += dice.item()
                val_batches += 1
        
        # Вычисляем средние значения
        avg_train_total = train_total_loss / train_batches
        avg_val_total = val_total_loss / val_batches
        avg_val_dice = val_dice / val_batches
        current_lr = optimizer.param_groups[0]['lr']
        
        epoch_time = time.time() - epoch_start
        
        print(f"✅ Эпоха {epoch+1} завершена за {epoch_time:.1f}с")
        print(f"📊 Train Total: {avg_train_total:.4f} | Val Total: {avg_val_total:.4f}")
        print(f"🎯 Val Dice: {avg_val_dice:.4f} | LR: {current_lr:.2e}")
        
        # Визуализация прогресса
        if (epoch + 1) % 10 == 0 or epoch == 0:
            visualize_multi_task_progress(model, val_dataset, DEVICE, epoch+1)
        
        # Сохранение лучшей модели
        if avg_val_dice > best_dice:
            best_dice = avg_val_dice
            patience = 0
            torch.save({
                'epoch': epoch,
                'model_state_dict': model.state_dict(),
                'optimizer_state_dict': optimizer.state_dict(),
                'dice': best_dice,
            }, 'multi_task_vessel_model.pth')
            print(f"💾 Новая лучшая модель! Dice: {best_dice:.4f}")
        else:
            patience += 1
            print(f"⏳ Patience: {patience}/{max_patience}")
        
        scheduler.step()
        
        if patience >= max_patience:
            print(f"🛑 Ранняя остановка на эпохе {epoch+1}")
            break
    
    print(f"\n🎉 ОБУЧЕНИЕ ЗАВЕРШЕНО!")
    print(f"🏆 Лучший Dice: {best_dice:.4f}")

# =============================================
# ОБНОВЛЕННАЯ ВИЗУАЛИЗАЦИЯ
# =============================================

def visualize_multi_task_progress(model, dataset, device, epoch, num_samples=3):
    """Визуализация прогресса multi-task модели с ВСЕМИ bboxes"""
    model.eval()
    
    indices = random.sample(range(len(dataset)), num_samples)
    
    fig, axes = plt.subplots(num_samples, 4, figsize=(16, 4 * num_samples))
    if num_samples == 1:
        axes = [axes]
    
    for i, idx in enumerate(indices):
        with torch.no_grad():
            image, true_mask, true_bboxes = dataset[idx]
            seg_output, bbox_output = model(image.unsqueeze(0).to(device))
            
            pred_sigmoid = torch.sigmoid(seg_output)
            pred_binary = (pred_sigmoid > 0.3).float()
            
            # Bbox predictions
            pred_bboxes = bbox_output[0].cpu().numpy()  # [max_objects, 5]
            
        # Изображение
        img_np = image.cpu().permute(1, 2, 0).numpy()
        img_np = np.clip(img_np, 0, 1)
        
        axes[i][0].imshow(img_np)
        axes[i][0].set_title('Original Image')
        axes[i][0].axis('off')
        
        # True mask + ВСЕ bboxes
        true_np = true_mask.squeeze().cpu().numpy()
        axes[i][1].imshow(true_np, cmap='gray')
        
        # Рисуем ВСЕ true bboxes
        true_obj_count = 0
        for j in range(MAX_OBJECTS):
            if true_bboxes[j * 5 + 4] > 0.5:  # confidence > 0.5
                x_center, y_center, width, height = true_bboxes[j * 5:j * 5 + 4].numpy() * 256
                x1, y1 = x_center - width/2, y_center - height/2
                rect = plt.Rectangle((x1, y1), width, height, linewidth=2, edgecolor='green', facecolor='none')
                axes[i][1].add_patch(rect)
                axes[i][1].text(x1, y1 - 5, f'Obj{true_obj_count}', color='green', fontsize=8)
                true_obj_count += 1
        
        axes[i][1].set_title(f'True Mask + {true_obj_count} BBoxes')
        axes[i][1].axis('off')
        
        # Predicted mask
        pred_np = pred_binary.squeeze().cpu().numpy()
        axes[i][2].imshow(pred_np, cmap='gray')
        axes[i][2].set_title('Predicted Mask')
        axes[i][2].axis('off')
        
        # Predicted mask + ВСЕ bboxes
        axes[i][3].imshow(pred_np, cmap='gray')
        
        # Рисуем ВСЕ predicted bboxes
        pred_obj_count = 0
        for j in range(MAX_OBJECTS):
            confidence = 1 / (1 + np.exp(-pred_bboxes[j, 4]))  # sigmoid для confidence
            if confidence > 0.3:  # confidence > 0.3
                x_center, y_center, width, height = pred_bboxes[j, :4] * 256
                x1, y1 = x_center - width/2, y_center - height/2
                rect = plt.Rectangle((x1, y1), width, height, linewidth=2, edgecolor='red', facecolor='none')
                axes[i][3].add_patch(rect)
                axes[i][3].text(x1, y1 - 5, f'Obj{pred_obj_count}\n{confidence:.2f}', color='red', fontsize=8)
                pred_obj_count += 1
        
        axes[i][3].set_title(f'Pred Mask + {pred_obj_count} BBoxes')
        axes[i][3].axis('off')
    
    plt.suptitle(f'Epoch {epoch} - Multi-Task Training (ВСЕ объекты)', fontsize=16)
    plt.tight_layout()
    plt.savefig(f'multi_task_all_objects_epoch_{epoch}.png', dpi=100, bbox_inches='tight')
    plt.show()

# =============================================
# ЗАПУСК
# =============================================

if __name__ == "__main__":
    print("⚡ MULTI-TASK VESSEL TRAINING - ВСЕ ОБЪЕКТЫ!")
    print("=" * 60)
    
    main()
```
